```python
import numpy as np
#numpY: libreria para trabajar convectores , matrices y operaciones numericas rapidas
import pandas as pd
#pandas: manejo de dataframes (tablas tipo excel)
import matplotlib.pyplot as plt
import seaborn as sns
#matplotlib :para graficas (lineas dispersion histogramas etc)
from sklearn.model_selection import train_test_split
#de scokit-learn : sirve para dividir datos de entrenamiento y prueba
from sklearn.linear_model import LinearRegression
#modelo de regresion lineal de scikit-learn
from sklearn.metrics import mean_squared_error, r2_score , mean_absolute_error
#metricas de evaluacion de modelos :
#MSE(mean squared error) error cuadratico medio
#MAE(mean absolute error) error absolito medio
#R^2 (coeficiente de determinacion) mide que tan bien explica el modelos la variabilidad
```

```python
#usaremos ´age´ como predictor (x) y ´fare´ como variable dependiente (y)
#eliminamos filas y valores nulos
df = df[["Age","Fare"]].dropna()
```

```python
X = df[["Age"]] #variable independiente
Y = df["Fare"] #variable dependiente
```

```python
#generar analisis
modelo = LinearRegression()
modelo.fit (X, Y)
```

```python
print ('Ecuación de la recta: y = ', round(modelo.coef_[0],3),'x + ', round(modelo.intercept_,3))
#obtenemos la ecuación de la recta
print ('Coeficiente de correlación: ', round(np.corrcoef(df['Age'], df['Fare'])[0,1], 3))
#obtenemos el coeficiente de correlación
print ('Coeficiente de determinación: ', round(r2_score(Y, modelo.predict(X)), 3))
#obtenemos el coeficiente de determinación
# Gráfica con intervalo de confianza
sns.regplot(x=df["Age"], y=df["Fare"], ci=95, line_kws={"color": "red"}, scatter_kws={"color": "black"})
plt.title('Regresión lineal simple con intervalo de confianza al 95%')
plt.show()
```
